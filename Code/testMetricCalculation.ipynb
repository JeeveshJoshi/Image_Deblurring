This code is used to compare the original sharp image, with our generated deblurred image to calculate how successful we were.

```python
import os
import torch
import lpips
import numpy as np
from skimage.io import imread
from skimage.metrics import peak_signal_noise_ratio as compare_psnr
from skimage.metrics import structural_similarity as compare_ssim
from tqdm import tqdm

lpips_model = lpips.LPIPS(net='alex').eval()

psnr_scores = []
ssim_scores = []
lpips_scores = []


filename = 'blurred-image.jpeg'
print(filename)

deblurredImage = imread('./../Dataset/ValidationDataset/tree_deblur.png')/ 255.0
sharpImage = imread('./../Dataset/ValidationDataset/tree_sharp.png')/ 255.0


if deblurredImage.shape != sharpImage.shape:
    print(f"Skipping {filename}: size mismatch.")
    pass

# Calculate Peak Signal to Noise Ratio
psnr = compare_psnr(sharpImage, deblurredImage, data_range=1.0)
psnr_scores.append(psnr)

# Calculate Structural Similarity Index Measure
ssim = compare_ssim(sharpImage, deblurredImage, channel_axis=-1, data_range=1.0)
ssim_scores.append(ssim)

# Calculate Learned Perceptual Image Patch Similarity
tensor_d = torch.tensor(deblurredImage).permute(2, 0, 1).unsqueeze(0).float()
tensor_g = torch.tensor(sharpImage).permute(2, 0, 1).unsqueeze(0).float()
lpips_score = lpips_model(tensor_d, tensor_g).item()
lpips_scores.append(lpips_score)

# Results
print("\n--- Average Metrics ---")
print(f"PSNR:  {np.mean(psnr_scores):.2f} dB")
print(f"SSIM:  {np.mean(ssim_scores):.4f}")
print(f"LPIPS: {np.mean(lpips_scores):.4f}")
```
